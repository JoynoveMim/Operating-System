```python
process = ["P1", "P2", "P3", "P4", "P5", "P6"]
burst = [7, 4, 15, 11, 20, 9]
Arrive = [0, 1, 2, 3, 4, 4]

quantum = 2

remaining = burst.copy()
completion = [0] * 6

time = 0

while True:
    done = True

    for i in range(6):

        # Process has arrived and still has work
        if Arrive[i] <= time and remaining[i] > 0:

            done = False

            if remaining[i] > quantum:
                time = time + quantum
                remaining[i] = remaining[i] - quantum

            else:
                time = time + remaining[i]
                remaining[i] = 0
                completion[i] = time

                print(process[i], "completed at", time)

    if done:
        break
```
