```python
request = [0, 41, 30, 100, 62, 51, 20]
head = 70
direction = "left"

left = []
right = []

for r in request:
    if r < head:
        left.append(r)
    else:
        right.append(r)

left.sort(reverse=True)
right.sort()

order = [head]

if direction == "left":
    for r in left:
        order.append(r)

    for r in right:
        order.append(r)

total_movement = 0

for i in range(len(order) - 1):
    total_movement += abs(order[i + 1] - order[i])

print("Order:", order)
print("Total Head Movement =", total_movement)
```

```python
request = [0, 41, 30, 100, 62, 51, 20]
head = 70
total_movement = 0

print("Order:", head, end="")

for r in request:
    total_movement += abs(r - head)
    head = r
    print(" ->", head, end="")

print("\nTotal Head Movement =", total_movement)
```
