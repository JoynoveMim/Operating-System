```python




# Priority Scheduling - Low to High

processes = [
    ["P1", 0, 3, 4],
    ["P2", 2, 2, 3],
    ["P3", 1, 1, 1],
    ["P4", 4, 2, 0],
    ["P5", 3, 2, 2],
    ["P6", 5, 3, 2]
]

n = len(processes)

completed = [False] * n
completion_time = [0] * n
turnaround_time = [0] * n
waiting_time = [0] * n

current_time = 0
completed_count = 0


while completed_count < n:

    index = -1
    highest_priority = 9999

    for i in range(n):
        if (not completed[i] and
            processes[i][1] <= current_time and
            processes[i][3] < highest_priority):

            highest_priority = processes[i][3]
            index = i

    if index == -1:
        current_time += 1
        continue

    pid = processes[index][0]
    at = processes[index][1]
    bt = processes[index][2]

    start_time = current_time
    current_time += bt

    completion_time[index] = current_time
    turnaround_time[index] = completion_time[index] - at
    waiting_time[index] = turnaround_time[index] - bt

    completed[index] = True
    completed_count += 1

    gantt.append((pid, start_time, current_time))



print("PID\tAT\tBT\tPriority\tCT\tTAT\tWT")

for i in range(n):
    print(
        processes[i][0], "\t",
        processes[i][1], "\t",
        processes[i][2], "\t",
        processes[i][3], "\t\t",
        completion_time[i], "\t",
        turnaround_time[i], "\t",
        waiting_time[i]
    )



avg_wt = sum(waiting_time) / n

avg_tat = sum(turnaround_time) / n

print("\nAverage Waiting Time =", avg_wt)
print("Average Turnaround Time =", avg_tat)

print()

```

```python
# Higher value = Higher priority

processes = [
    ["P1", 0, 3, 4],
    ["P2", 2, 2, 3],
    ["P3", 1, 1, 1],
    ["P4", 4, 2, 0],
    ["P5", 3, 2, 2],
    ["P6", 5, 3, 2]
]

n = len(processes)

completed = [False] * n
CT = [0] * n
TAT = [0] * n
WT = [0] * n

current_time = 0
completed_count = 0

while completed_count < n:

    index = -1
    highest_priority = -1

    # Find highest priority process
    for i in range(n):
        if (not completed[i] and
            processes[i][1] <= current_time and
            processes[i][3] > highest_priority):

            highest_priority = processes[i][3]
            index = i

    # If no process has arrived
    if index == -1:
        current_time += 1
        continue

    pid = processes[index][0]
    AT = processes[index][1]
    BT = processes[index][2]

    start_time = current_time
    current_time += BT

    CT[index] = current_time
    TAT[index] = CT[index] - AT
    WT[index] = TAT[index] - BT

    completed[index] = True
    completed_count += 1

    gantt.append((pid, start_time, current_time))


# Display table
print("PID\tAT\tBT\tPriority\tCT\tTAT\tWT")

for i in range(n):
    print(
        processes[i][0], "\t",
        processes[i][1], "\t",
        processes[i][2], "\t",
        processes[i][3], "\t\t",
        CT[i], "\t",
        TAT[i], "\t",
        WT[i]
    )

# Average
print("\nAverage Waiting Time =", sum(WT) / n)
print("Average Turnaround Time =", sum(TAT) / n)

print()
```
