```python
processId = ["p4", "p5", "p2", "p1", "p3"]
arrivetime = [0, 1, 2, 3, 5]
burst = [3, 2, 1, 3, 2]

time = 0

waitingtime = []
comtime = []
tatime = []

for i in range(len(burst)):

    waiting = time - arrivetime[i]
    waitingtime.append(waiting)

    time = time + burst[i]
    comtime.append(time)

    tat = time - arrivetime[i]
    tatime.append(tat)

print("Waiting Time =", waitingtime)
print("Completion Time =", comtime)
print("Turnaround Time =", tatime)
```
