```python
p = [
    ["P1", 3, 3, 1],
    ["P2", 0, 4, 3],
    ["P3", 2, 2, 1],
    ["P4", 1, 3, 1],
    ["P5", 2, 4, 0]
]

n = len(p)
remaining = [p[i][2] for i in range(n)]

CT = [0] * n
TAT = [0] * n
WT = [0] * n

time = 0
completed = 0
quantum = 1

gantt = []
current = None
current_start = 0

while completed < n:
    best = -1
    best_pri = float('inf')

    for i in range(n):
        if p[i][1] <= time and remaining[i] > 0:
            if p[i][3] < best_pri:
                best_pri = p[i][3]
                best = i
            elif p[i][3] == best_pri:

                if current is not None and i == current:
                    best = i

                elif current is not None and best == current:
                    pass
                else:

                    i_started = remaining[i] < p[i][2]
                    b_started = remaining[best] < p[best][2]
                    if i_started and not b_started:
                        best = i
                    elif i_started == b_started:

                        if p[i][0] < p[best][0]:
                            best = i


    if best == -1:
        if current is not None:
            gantt.append((p[current][0], current_start, time))
            current = None
        time += 1
        continue


    if best != current:
        if current is not None:
            gantt.append((p[current][0], current_start, time))
        current = best
        current_start = time


    remaining[current] -= 1
    time += 1

    if remaining[current] == 0:
        CT[current] = time
        completed += 1
        gantt.append((p[current][0], current_start, time))
        current = None


for i in range(n):
    TAT[i] = CT[i] - p[i][1]
    WT[i] = TAT[i] - p[i][2]


print("Process\tAT\tBT\tPriority\tCT\tTAT\tWT")
for i in range(n):
    print(f"{p[i][0]}\t{p[i][1]}\t{p[i][2]}\t{p[i][3]}\t\t{CT[i]}\t{TAT[i]}\t{WT[i]}")

print("\nAverage TAT =", sum(TAT) / n)
print("Average WT  =", sum(WT) / n)


```

```python
p = [
    ["P1", 3, 3, 1],
    ["P2", 0, 4, 3],
    ["P3", 2, 2, 2],
    ["P4", 1, 3, 1],
    ["P5", 2, 4, 0]
]


quantum = 3
n = len(p)

remaining = {x[0]: x[2] for x in p}
arrival = {x[0]: x[1] for x in p}
priority = {x[0]: x[3] for x in p}

time = 0
completed = 0
gantt = []
completion = {}

while completed < n:


    ready = [
        x for x in p
        if arrival[x[0]] <= time and remaining[x[0]] > 0
    ]

    if not ready:
        gantt.append(("Idle", time, time + 1))
        time += 1
        continue


    ready.sort(key=lambda x: (priority[x[0]], arrival[x[0]]))

    current = ready[0][0]

    run_time = min(quantum, remaining[current])

    start = time
    time += run_time
    remaining[current] -= run_time

    gantt.append((current, start, time))

    if remaining[current] == 0:
        completed += 1
        completion[current] = time


turnaround = {}
waiting = {}

for x in p:
    name = x[0]
    at = x[1]
    bt = x[2]

    ct = completion[name]
    tat = ct - at
    wt = tat - bt

    turnaround[name] = tat
    waiting[name] = wt

print("\nProcess\tAT\tBT\tPriority\tCT\tTAT\tWT")

for x in p:
    name, at, bt, pr = x

    print(
        f"{name}\t{at}\t{bt}\t{pr}\t\t"
        f"{completion[name]}\t"
        f"{turnaround[name]}\t"
        f"{waiting[name]}"
    )






avg_wt = sum(waiting.values()) / n
avg_tat = sum(turnaround.values()) / n

print("\nAverage Waiting Time =", avg_wt)
print("Average Turnaround Time =", avg_tat)

```
